```python
import requests
base = "https://data.cityofnewyork.us/resource/{}.json"
for name, ds in [("persons", "f55k-p6yu"), ("vehicles", "bm4k-52h4")]:
    rows = requests.get(base.format(ds), params={"$select": "collision_id", "$limit": 3}).json()
    print(name, [r.get("collision_id") for r in rows])
```
